```python
df1['year'].value_counts()
```

```python
# Extract only 4-digit numbers from the column
df['year'] = df['year'].str.extract(r'(\d{4})')
```

```python
df1['year']
```

```python
# Convert to numeric
#✅ What to_numeric() does
#pd.to_numeric(..., errors='coerce') converts:
# "2007" → 2007.0
# "abcd" → NaN
# "r 15" → NaN
# "2005" → 2005.0
# So now all valid numbers become numeric (float), and junk becomes NaN.
# Then we drop NaN, and only THEN convert to int.

df1['year'] = pd.to_numeric(df1['year'], errors='coerce')
```

```python
# keep only reasonable year range
df1['year'] = df1['year'].where(df1['year'].between(1990, 2025))
```

```python
# Drop rows where year is missing after cleaning
df1 = df1.dropna(subset=['year'])

# Convert to int
df1['year'] = df1['year'].astype(int)
```

```python
df['Price']
```

```python
df1 = df1[df1['Price'].str.contains(r'^[0-9, ]+$', na=False)]
```

```python
df1['Price'] = df1['Price'].str.replace(',', '', regex=False).str.replace(' ', '')
```

```python
df1['Price'] = df1['Price'].astype(int)
```

```python
df1.info()
```

```python
df1['kms_driven']
```

```python
# 1. Remove the word 'kms' (case-insensitive)
df1['kms_driven'] = df1['kms_driven'].str.replace('kms', '', case=False, regex=False)
```

```python
# 2. Remove commas and spaces
df1['kms_driven'] = df1['kms_driven'].str.replace(',', '', regex=False).str.replace(' ', '')
```

```python
df1 = df1[df1['kms_driven'].str.contains(r'^\d+$', na=False)]
```

```python
df1['kms_driven'] = df1['kms_driven'].astype(int)
```

```python
df1.info()
```

```python
df1 = df1.dropna(subset=['fuel_type'])
```
